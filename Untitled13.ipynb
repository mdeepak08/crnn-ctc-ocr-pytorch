```python
import csv, random
from pathlib import Path
from io import BytesIO

from datasets import load_dataset, Image as HFImage
from PIL import Image

random.seed(42)

# Where to store images/CSVs
out_raw = Path("data/raw/mjsynth/images")
out_raw.mkdir(parents=True, exist_ok=True)
out_proc = Path("data/processed")
out_proc.mkdir(parents=True, exist_ok=True)

# Start small; increase later
N = 200_000
VAL_RATIO = 0.05

allowed = set("0123456789abcdefghijklmnopqrstuvwxyz")
max_len = 25

ds = load_dataset("priyank-m/MJSynth_text_recognition", split="train", streaming=True)
ds = ds.cast_column("image", HFImage(decode=False))  # avoids decode crashes

# Auto-detect label key
TEXT_KEYS = ["text", "label", "word", "transcription", "gt", "ground_truth"]

def get_text(ex):
    for k in TEXT_KEYS:
        if k in ex and ex[k] is not None:
            return str(ex[k]).strip()
    return ""

rows = []
bad_img = 0
bad_text = 0
scanned = 0

for ex in ds:
    scanned += 1

    text = get_text(ex).lower()
    if (not text) or (len(text) > max_len) or any(c not in allowed for c in text):
        bad_text += 1
        continue

    img_obj = ex.get("image", None)
    try:
        if isinstance(img_obj, dict) and img_obj.get("bytes") is not None:
            img = Image.open(BytesIO(img_obj["bytes"]))
        elif isinstance(img_obj, dict) and img_obj.get("path"):
            img = Image.open(img_obj["path"])
        else:
            bad_img += 1

        img = img.convert("RGB")
        p = out_raw / f"{len(rows):08d}.jpg"
        img.save(p, quality=95)

        rows.append((str(p), text))
        if len(rows) >= N:
            break

    except Exception:
        bad_img += 1
        continue

    if len(rows) and len(rows) % 20000 == 0:
        print(f"kept={len(rows)} scanned={scanned} bad_text={bad_text} bad_img={bad_img}")

print(f"FINAL kept={len(rows)} scanned={scanned} bad_text={bad_text} bad_img={bad_img}")

random.shuffle(rows)
n_val = max(1, int(len(rows) * VAL_RATIO))
val_rows = rows[:n_val]
train_rows = rows[n_val:]

def write_csv(path, rows_):
    with open(path, "w", newline="") as f:
        w = csv.writer(f)
        w.writerow(["path", "text"])
        w.writerows(rows_)

write_csv(out_proc / "mjsynth_train.csv", train_rows)
write_csv(out_proc / "mjsynth_val.csv", val_rows)

print("train:", len(train_rows), "val:", len(val_rows))
print("example:", train_rows[0] if train_rows else "NO TRAIN ROWS")
```

```python
%%bash
python3 - <<'PY'
import csv, shutil
from pathlib import Path

pred_csv = Path("eval_outputs/SVT/After_adding_rectifier/predictions (1).csv")

out_dir = Path("svt_result_images")
out_dir.mkdir(parents=True, exist_ok=True)

index_path = out_dir / "index.csv"
rows_out = []

n = 0
missing = 0

with pred_csv.open(newline="") as f:
    r = csv.DictReader(f)
    for row in r:
        if row.get("correct", "0") != "0":
            continue

        img_path = Path(row["path"])
        if not img_path.exists():
            missing += 1
            continue

        dst = out_dir / f"{n:04d}__{img_path.name}"
        shutil.copy2(img_path, dst)

        rows_out.append({
            "copied_name": dst.name,
            "orig_path": str(img_path),
            "gt": row.get("gt", ""),
            "pred": row.get("pred", ""),
            "correct": row.get("correct", ""),
        })
        n += 1

with index_path.open("w", newline="") as f:
    w = csv.DictWriter(f, fieldnames=["copied_name","orig_path","gt","pred","correct"])
    w.writeheader()
    w.writerows(rows_out)

print("copied wrong images:", n)
print("missing image files:", missing)
print("folder:", out_dir)
print("index:", index_path)
PY

zip -r svt_result_images.zip svt_result_images
```

```python
import csv, random
from pathlib import Path

# Input CSVs (edit if needed)
mjsynth_csv = Path("data/processed/mjsynth_train.csv")
iiit_csv    = Path("data/processed/iiit5k_train.csv")

# Output
out_csv      = Path("data/processed/mixed_mjsynth_iiit5k_train.csv")

# Optional sampling (set to None to keep all)
max_mjsynth = None        # e.g., 50000
max_iiit    = None        # usually keep all IIIT

seed = 42
random.seed(seed)

def read_rows(p: Path):
    with p.open(newline="") as f:
        r = csv.DictReader(f)
        assert "path" in r.fieldnames and "text" in r.fieldnames, (p, r.fieldnames)
        return [{"path": row["path"], "text": row["text"]} for row in r]

m = read_rows(mjsynth_csv)
i = read_rows(iiit_csv)

if max_mjsynth is not None and len(m) > max_mjsynth:
    m = random.sample(m, max_mjsynth)
if max_iiit is not None and len(i) > max_iiit:
    i = random.sample(i, max_iiit)

rows = m + i
random.shuffle(rows)

out_csv.parent.mkdir(parents=True, exist_ok=True)
with out_csv.open("w", newline="") as f:
    w = csv.DictWriter(f, fieldnames=["path", "text"])
    w.writeheader()
    w.writerows(rows)

print("wrote:", out_csv)
print("counts:", {"mjsynth": len(m), "iiit": len(i), "total": len(rows)})
```
